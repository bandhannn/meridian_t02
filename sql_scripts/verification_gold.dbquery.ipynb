```sql
USE CATALOG meridian_t02;

SELECT 
    'fct_sessions' AS fact_table,
    COUNT(*) AS total_rows,
    COUNT(CASE WHEN user_country IS NULL OR user_country = 'Unknown' THEN 1 END) AS missing_country_count
FROM gold.fct_sessions
UNION ALL
SELECT 
    'fct_bookings' AS fact_table,
    COUNT(*) AS total_rows,
    COUNT(CASE WHEN user_country IS NULL OR user_country = 'Unknown' THEN 1 END) AS missing_country_count
FROM gold.fct_bookings
UNION ALL
SELECT 
    'fct_reviews' AS fact_table,
    COUNT(*) AS total_rows,
    COUNT(CASE WHEN user_country IS NULL OR user_country = 'Unknown' THEN 1 END) AS missing_country_count
FROM gold.fct_reviews;
```
