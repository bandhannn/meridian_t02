```sql
-- Profile Clickstream: event types, timestamp bounds, user coverage, and device struct
SELECT 
    COUNT(*) AS total_records,
    COUNT(DISTINCT user_id) AS unique_users,
    MIN(timestamp) AS min_event_time,
    MAX(timestamp) AS max_event_time,
    COUNT(CASE WHEN user_id IS NULL THEN 1 END) AS null_user_ids,
    COUNT(CASE WHEN property_id IS NULL THEN 1 END) AS null_property_ids
FROM samples.wanderbricks.clickstream;

SELECT event, metadata.device, metadata.referrer, COUNT(*) AS event_count
FROM samples.wanderbricks.clickstream
GROUP BY event, metadata.device, metadata.referrer
ORDER BY event_count DESC;

-- Profile Users: check country distribution (RLS dimension) and user types (SCD2)
SELECT 
    COUNT(*) AS total_users,
    COUNT(DISTINCT user_id) AS unique_user_ids,
    COUNT(DISTINCT email) AS unique_emails,
    COUNT(CASE WHEN country IS NULL THEN 1 END) AS null_countries,
    user_type,
    country,
    COUNT(*) AS count_per_type_country
FROM samples.wanderbricks.users
GROUP BY user_type, country;

-- Profile Reviews: ratings distribution and soft-deletion flag
SELECT 
    COUNT(*) AS total_reviews,
    COUNT(DISTINCT review_id) AS unique_reviews,
    MIN(rating) AS min_rating,
    MAX(rating) AS max_rating,
    COUNT(CASE WHEN is_deleted = TRUE THEN 1 END) AS deleted_reviews_count,
    COUNT(CASE WHEN comment IS NULL THEN 1 END) AS null_comments
FROM samples.wanderbricks.reviews;

-- Profile Bookings: conversion target dates and totals
SELECT 
    COUNT(*) AS total_bookings,
    COUNT(DISTINCT booking_id) AS unique_bookings,
    MIN(check_in) AS min_checkin,
    MAX(check_out) AS max_checkout,
    MIN(total_amount) AS min_amount,
    MAX(total_amount) AS max_amount,
    status,
    COUNT(*) AS count_by_status
FROM samples.wanderbricks.bookings
GROUP BY status;

-- Profile Properties: types and price points
SELECT 
    COUNT(*) AS total_properties,
    COUNT(DISTINCT property_id) AS unique_properties,
    property_type,
    MIN(base_price) AS min_price,
    MAX(base_price) AS max_price,
    COUNT(*) AS property_count
FROM samples.wanderbricks.properties
GROUP BY property_type;
```
