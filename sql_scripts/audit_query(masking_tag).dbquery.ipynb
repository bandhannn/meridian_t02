```sql
USE CATALOG meridian_t02;

-- Audit Query: Match tagged sensitive columns with active mask policies
SELECT 
    t.schema_name,
    t.table_name,
    t.column_name,
    t.tag_name,
    t.tag_value,
    m.mask_name,
    CASE 
        WHEN m.mask_name IS NOT NULL THEN 'COMPLIANT (MASKED)'
        ELSE 'NON-COMPLIANT (UNMASKED)'
    END AS governance_status
FROM system.information_schema.column_tags t
LEFT JOIN system.information_schema.column_masks m
  ON t.catalog_name = m.table_catalog
 AND t.schema_name  = m.table_schema
 AND t.table_name   = m.table_name
 AND t.column_name  = m.column_name
WHERE t.catalog_name = 'meridian_t02'
  AND t.tag_name     = 'contains_pii' 
  AND t.tag_value    = 'true';
```
